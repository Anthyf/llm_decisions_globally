```python
import pickle
from pathlib import Path
from IPython.display import display
import numpy as np

path_results = Path("results_global/country/test_25dollar_fit101/data_log.pkl")
# results_global/country/test_10countries/data_log.pkl

with open(path_results, "rb") as f:
    data = pickle.load(f)

# print (data)
# check for each model, how many unique outputs from it.
for i in data.keys():
    print( f"Model: {i}" )
    data_model=data[i]
    data_model=data_model[data_model['reward_value']==40]
    values, counts = np.unique(data_model['output'], return_counts=True)
    print(f"Unique values: {values}")
    print(f"Counts: {counts}")
    data_model=data[i].sort_values(by=["country","reward_value"], ascending=[True,True]) 
    print(data_model)
```
